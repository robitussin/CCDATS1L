```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
```

```python
x = [92,145,130,70,75,105,110,108,45,50,160,155,180,190,63,85,130,132]
y = [6.3, 7.8, 3, 5.5, 6.5, 5.5, 6.5, 8, 4, 5, 7.5, 9, 8.6, 10, 4.2, 4.9, 6,7]
```

```python
df = pd.DataFrame({'hours_studied': x, 'exam_score': y})
df.head(100)
```

```python
df.plot.scatter(x='hours_studied', y='exam_score', title='Scatter plot between hours studied and exam score')
```

```python
correlation = df.corr()
```

```python


plt.figure(figsize=(8, 6)) # Optional: Adjust the figure size
sns.heatmap(correlation, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.show() # Display the plot
```

```python
# Calculate for two columns
corr, p_value = pearsonr(df['hours_studied'], df['exam_score'])

print(f"Correlation: {corr:.4f}")
print(f"P-value: {p_value:.4f}")

```

```python
alpha = 0.05  # significance level

if p_value < alpha:
    print("Reject H₀: There is a statistically significant correlation.")
else:
    print("Fail to reject H₀: No statistically significant correlation.")
```
