```python
from scipy import stats

# Sample data for three different groups
group_a = [85, 86, 88, 75, 78, 94, 98, 79, 71, 80]
group_b = [91, 92, 93, 85, 87, 84, 82, 88, 95, 96]
group_c = [79, 78, 88, 94, 92, 85, 83, 85, 82, 81]

# Perform One-way ANOVA
f_statistic, p_value = stats.f_oneway(group_a, group_b, group_c)

print(f"F-statistic: {f_statistic:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Significant difference found (Reject H0)")
else:
    print("No significant difference found (Fail to reject H0)")
```
